# 回测
## 交易规则
- 当量能指标数值高于设定阈值Threshold=1.15时，开仓做多万得全A指数；
- 当量能指标数值介于1与$threshold^{-a}$之间（参数$a$取值区间为$[1.5,3.5]$）时，开仓做空万得全A指数；
- 当量能指标数值低于$threshold^{-a}$时，代表市场进入地量反弹区间，后续执行做多操作。

```python
# 基于高斯过程实现策略参数贝叶斯优化
def optimize_strategy(
    dataset: pd.DataFrame, 
    index_name: str,
    random_seed: int = None, 
    **kwargs
):
    """
    策略参数自动优化函数
    通过最大化年化收益率筛选最优超参数组合
    """
    # 定义内部优化目标函数：计算策略年化收益并取负值（最小化问题）
    def execute_strategy_optimization(
        parameters, 
        input_data, 
        asset_name, 
        strategy_func
    ):
        # 解包待优化参数
        scale_coef, fast_period, slow_period = parameters
        
        # 约束条件：快速周期必须小于慢速周期
        if fast_period >= slow_period:
            return 0
        
        # 执行回测获取绩效结果
        backtest_output = get_backtesting(
            input_data,
            asset_name,
            strategy_func,
            a=scale_coef,
            fast_window=fast_period,
            slow_window=slow_period,
            print_log=False,
            **kwargs
        )

        # 返回负向年化收益率，适配最小化优化目标
        return -backtest_output[0][0].analyzers._Returns.get_analysis()["rnorm100"]

    # 启动贝叶斯优化流程
    optimization_result = gp_minimize(
        lambda params: execute_strategy_optimization(
            params, dataset, index_name, bimodal_distribution_strategy
        ),
        # 参数搜索空间：[系数, 快速窗口, 慢速窗口]
        [(1.5, 3.5), (5, 180), (5, 300)],
        x0=[1.5, 5, 10],
        n_calls=100,
        acq_func='EI',
        random_state=random_seed,
    )

    return optimization_result

# ===================== 样本内参数优化执行 =====================
# 设定样本内回测时间区间
sample_start, sample_end = "2004-01-01", "2019-12-31"

# 定义需要筛选的核心行情字段
target_columns: list = ["close", "open", "high", "low", "volume"]

# 筛选指定指数与时间区间的数据
filtered_data: pd.DataFrame = data.query(
    'code=="881001.WI"'
).loc[sample_start:sample_end, target_columns]

# 剔除缺失值，保证回测数据完整性
filtered_data = filtered_data.dropna()

# 执行参数优化，获取最优参数结果
optimal_parameters = optimize_strategy(filtered_data, "wind_A")
```

```python
# 策略回测执行模块
# 解析贝叶斯优化输出的最优参数组合
backtest_parameters: Dict = dict(zip(["a", "fast_window", "slow_window"], optimal_parameters.x))
# 控制台打印最优参数结果，便于记录与核对
print(f"最优参数:{backtest_parameters}")

# 执行完整策略回测
# trade_long=False 代表开启多空交易模式；trade_long=True 代表仅做多
backtest_result = get_backtesting(
    filtered_data,
    "wind_A",
    bimodal_distribution_strategy,
    **backtest_parameters,
    trade_long=False,
    print_log=False,
)
```

## 宽基指数测试

```python
# 批量回测执行函数：对多只标的同时进行策略回测
def batch_backtest_executor(
    dataset: pd.DataFrame, 
    parameters: Dict, 
    start_date: str = None, 
    end_date: str = None
) -> Dict:
    """
    批量执行策略回测，支持多标的、自定义时间区间
    返回：标的代码 -> 回测结果 的字典
    """
    # 定义回测所需的核心行情字段
    required_columns: list = ["close", "open", "high", "low", "volume"]
    # 存储所有标的的回测结果
    backtest_result_dict: Dict = {}

    # 按标的代码分组遍历
    for ticker_code, ticker_data in dataset.groupby("code"):
        # 获取当前标的对应的策略参数
        strategy_kwargs = parameters[ticker_code]

        # 筛选指定时间区间的数据（如果传入时间参数）
        if start_date is not None and end_date is not None:
            processed_data = ticker_data.loc[start_date:end_date, required_columns].dropna()
        else:
            processed_data = ticker_data[required_columns].dropna()

        # 执行回测并捕获异常
        try:
            backtest_output = get_backtesting(
                processed_data,
                ticker_code,
                bimodal_distribution_strategy,
                print_log=False,
                **strategy_kwargs
            )
            backtest_result_dict[ticker_code] = backtest_output.result
        
        # 异常处理：打印报错标的并抛出异常
        except Exception as error:
            print(f"标的 {ticker_code} 回测执行失败")
            raise error

    return backtest_result_dict
```

```python
# 批量生成信号并绘制各标的特征分布图表
# 遍历所有指数代码，逐一生成交易信号并可视化特征分布
for index_code, index_data in data.groupby("code"):
    
    # 为当前标的生成标准化信号数据
    signal_dataset = create_signal(
        data.query(f'code=="{index_code}"'),
        fast_window=5,
        slow_window=100,
        threshold=1.15,
        a=1.5,
    )

    # 绘制成交量指标与未来收益率的分布关系图
    plot_distribution(
        signal_dataset["volume_index"],
        signal_dataset["forward_returns"],
        5,
        50,
        False,
        title=index_code
    )
```

使用opt函数对每个指数进行寻参,并进行回测

```python
# 批量参数寻优 + 全品种回测
sample_start, sample_end = "2004-01-01", "2019-12-31"

# 定义回测必需的核心行情字段
target_columns: List = ["close", "open", "high", "low", "volume"]

# 遍历所有标的，通过贝叶斯优化自动搜索最优参数
optimized_params: Dict = {
    code: dict(
        zip(
            ["a", "fast_window", "slow_window"],
            optimize_strategy(
                dataset.loc[sample_start:sample_end, target_columns].dropna(),
                code
            ).x,
        )
    )
    for code, dataset in data.groupby("code")
}

# 输出所有品种的最优参数
print(f"寻参结果如下:{optimized_params}")

# 执行批量回测，使用优化后的参数组合
batch_backtest_result: Dict = batch_backtest_executor(
    data, optimized_params, sample_start, sample_end
)
```

```python
# 根据优化拟合的参数绘制策略累计收益曲线
result_count = len(batch_backtest_result)
column_num = 2
row_num = result_count // 2 + 1 if result_count % 2 != 0 else result_count // 2

# 初始化网格画布
grid_plot = GridFigure(row_num, column_num)
grid_plot.fig = plt.figure(figsize=(14, row_num * 4))

# 遍历所有标的回测结果，绘制收益曲线
for code, bt_output in batch_backtest_result.items():
    # 计算策略累计收益率
    cumulative_returns = ep.cum_returns(get_time_returns(bt_output[0]))
    # 在子图中绘制收益曲线并设置标题
    cumulative_returns.plot(title=code, color="red", ax=grid_plot.next_cell())

plt.show()
grid_plot.close()
```
