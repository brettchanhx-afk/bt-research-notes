# 价量共振
## 价量共振指标择时体系构建
1. 成交量维度选取AMA均线，收盘价维度选取BMA均线，均线参数长度统一设定为L；
2. 价能指标计算方式为当日BMA数值与前N个交易日BMA数值的比值：$\frac{BMA_{today}}{BMA_{Today-N}}$；
3. 量能指标计算方式为5日AMA数值与长期AMA数值的比值：$\frac{AMA_{5}}{AMA_{Long}}$；
4. 价量共振指标由价能与量能二者相乘得到；
5. 若价量共振指标数值高于设定阈值，则开仓做多，反之进行平仓操作。

## 基于市场状态划分的择时体系
1. 以5日均线与90日均线相对大小划分市场行情：5日均线高于90日均线判定为多头行情，5日均线低于90日均线判定为空头行情；
2. 处于多头行情阶段时，价量共振指标高于Threshold1则开仓做多，反之依据Threshold1执行平仓；处于空头行情阶段时，价量共振指标高于Threshold2则开仓做多，反之依据Threshold2执行平仓。

**参数设定**
AMA长期均线周期为100，BMA均线周期为50，参数N取值为3，Threshold1取值为1.125，Threshold2取值为1.275。

````python
# 获取股票指数分类映射数据
index_category_map = fetch_stock_index_mapping()

# 指定目标指数代码：沪深300
target_index_code = '000300.SH'

# 加载指数历史行情数据
hs300_index_data = fetch_market_data(
    target_index_code,
    "2010-01-01",
    "2022-10-11",
    fields=["close", "low", "high", "open", "volume"],
    method="index"
)

# 将交易日期设置为数据框索引
hs300_index_data.set_index("trade_date", inplace=True)
````

````python
# 基于Backtrader回测框架执行策略回测
backtest_result = run_backtest_engine(
    hs300_index_data,
    begin_dt='2010-01-01',
    end_dt='2022-06-14',
    name=index_category_map[target_index_code]
)
````

````python
# 基于指数收盘价与回测结果生成收益分析报告
performance_report = evaluate_return_indicators(
    hs300_index_data['close'],
    backtest_result[0]
)
````

````python
# 遍历收益分析报告中的所有图表并展示
for chart in performance_report:
    chart.show()
````

````python
# 基于收盘价与回测结果生成交易行为分析报告
trade_analysis_report = evaluate_trade_details(
    hs300_index_data['close'],
    backtest_result[0]
)
````

````python
# 遍历交易分析报告中的全部可视化图表并进行展示
for trade_chart in trade_analysis_report:
    trade_chart.show()
````

# 申万行业

````python
# 导入自定义信号计算核心模块
from source.signal_generator import batch_calculate_signal, fetch_signal_state_info
# 导入自定义网格可视化绘图模块
from source.plotter import MultiChartPlotly
# 导入全局回测配置与数据格式转换工具
from source.utils import GLOBAL_BACKTEST_SETTING, convert_state_dataframe

# 获取申万行业分类映射关系数据
sw_industry_category = fetch_sw_industry_classify()

# 批量获取申万行业指数的历史行情数据
sw_industry_price_data = fetch_market_data(
    list(sw_industry_category.keys()),
    "2010-01-01",
    "2022-10-11",
    fields=["close", "low", "high", "open", "volume"],
    method="sw"
)

# 将交易日期字段设置为数据索引
sw_industry_price_data.set_index("trade_date", inplace=True)

# 批量计算行业交易信号序列
trade_signal_series = batch_calculate_signal(
    sw_industry_price_data,
    **GLOBAL_BACKTEST_SETTING,
    level='sw',
    method='flag'
)

# 批量计算行业量能动量指标序列
volume_momentum_series = batch_calculate_signal(
    sw_industry_price_data,
    **GLOBAL_BACKTEST_SETTING,
    level='sw',
    method='vol_mom'
)
````

````python
# 按时间维度分组统计信号整体状态
signal_status_summary = trade_signal_series.groupby(level=0).apply(fetch_signal_state_info)

# 转换信号状态表格式并输出
convert_state_dataframe(signal_status_summary)
````

## 申万行业指数择时情况

````python
# 为行情数据添加行业名称标签，格式化显示内容
sw_industry_price_data['sec_name'] = sw_industry_price_data['code'].apply(
    lambda x: f"{sw_industry_category[x].replace('(申万)', '')}({x})"
)

# 构建收盘价透视表，行为交易日期、列为行业名称
close_price_table = pd.pivot_table(
    sw_industry_price_data.reset_index(),
    index='trade_date',
    columns='sec_name',
    values='close'
)

# 计算基准日收益率
benchmark_daily_return = close_price_table.pct_change()

# 计算策略组合日收益率：信号滞后一期与基准收益率相乘
strategy_daily_return = trade_signal_series.unstack(level=0).shift(1) * benchmark_daily_return

# 计算策略累计收益率
strategy_cumulative_return = empyrical_stats.cum_returns(strategy_daily_return)

# 计算基准累计收益率
benchmark_cumulative_return = empyrical_stats.cum_returns(benchmark_daily_return)
````

````python
# 调用网格绘图工具，展示策略与基准的累计收益对比曲线
MultiChartPlotly(
    strategy_cumulative_return,
    benchmark_cumulative_return
)
````

````python
# 计算最新交易日量价共振因子得分并降序排序
momentum_score = volume_momentum_series.unstack(level=0).iloc[-1].sort_values(ascending=False)

# 创建柱状图可视化因子得分分布
momentum_figure = plotly_fig.Figure(
    data=[plotly_fig.Bar(
        y=momentum_score.values,
        x=momentum_score.index,
        marker_color='crimson'
    )]
)

# 配置图表标题与字体样式
momentum_figure.update_layout(
    title=dict(
        text='量价共振因子得分排行',
        font={"size": 30}
    )
)
````
