## 3. 久期风格分析

```python
# ============================================================
# 研报定义: 加权平均久期 = Σ(W_i * D_i) / ΣW_i
# ============================================================

# 构建久期和评级字典
durations = {row['bond_code']: row['duration'] for _, row in holdings_df.iterrows()}
ratings   = {row['bond_code']: row['rating']   for _, row in holdings_df.iterrows()}

factor = BondStyleFactor()

# 计算久期风格
dur_style, dur_label = factor.calc_duration_style(
    holdings_df,
    durations=durations,
    weight_col='market_value'
)

print('='*50)
print(f'  久期风格分析结果')
print('='*50)
print(f'  加权平均久期: {dur_style:.2f} 年')
print(f'  风格标签: {dur_label} (期)')
print('='*50)

# 显示持仓久期分布nholdings_df[['bond_name', 'market_value', 'duration']].sort_values('market_value', ascending=False)
```

## 4. 信用风格分析

```python
# ============================================================
# 研报定义: 加权平均信用评分 = Σ(W_i * C_i) / ΣW_i
# 评分表 (来源: 人民银行 2006 年规范):
#   AAA: 16.5 | AA: 15.0 | A: 12.5 | BBB: 10.5 | BB: 8.5 | B: 6.5
# ============================================================

cred_style, cred_label = factor.calc_credit_style(
    holdings_df,
    ratings=ratings,
    weight_col='market_value'
)

print('='*50)
print(f'  信用风格分析结果')
print('='*50)
print(f'  加权平均信用评分: {cred_style:.2f} 分')
print(f'  风格标签: {cred_label} (等级)')
print('='*50)

# 显示评级分布
holdings_df[['bond_name', 'market_value', 'rating']].sort_values('market_value', ascending=False)
```

## 6. 多期滚动跟踪

```python
# ============================================================
# 演示多期风格测试 (实际项目通过 get_fund_holdings 获取历史数据)
# ============================================================

demo_records = []
periods = ['Q4-2023', 'Q3-2023', 'Q2-2023', 'Q1-2023']

for period in periods:
    # 每期指标略有所变化（演示用）
    np.random.seed(hash(period) % 2**32)
    dur_noise = np.random.normal(0, 0.3)
    cred_noise = np.random.normal(0, 0.2)

    demo_records.append({
        'period': period,
        'duration_style': max(0.1, result['duration_style'] + dur_noise),
        'credit_style': max(0.1, result['credit_style'] + cred_noise),
        'style_box': result['style_box'],
        'n_bonds': len(holdings_df),
    })

style_history = pd.DataFrame(demo_records)
print('\u98ce格测试历史:')
style_history
```

```python
# 绘制风格测试曲线
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 左图: 久期和信用分数时间序
ax1 = axes[0]
ax1.plot(range(len(style_history)), style_history['duration_style'], 'b-o', label='Duration (Y)', linewidth=2)
ax1.plot(range(len(style_history)), style_history['credit_style'],  'r-s', label='Credit Score', linewidth=2)
ax1.set_xticks(range(len(style_history)))
ax1.set_xticklabels(style_history['period'])
ax1.set_xlabel('Period')
ax1.set_ylabel('Score')
ax1.set_title('Style History')
ax1.legend()
ax1.grid(True, alpha=0.3)

# 右图: 久期分布
ax2 = axes[1]
colors = ['#A5D6A7' if d < 3.5 else '#FFF176' if d < 6 else '#EF9A9A'
          for d in style_history['duration_style']]
bars = ax2.bar(range(len(style_history)), style_history['duration_style'], color=colors, alpha=0.8)
ax2.axhline(y=3.5, color='gray', linestyle='--', linewidth=0.8, label='Short/Mid threshold')
ax2.axhline(y=6.0, color='gray', linestyle=':',  linewidth=0.8, label='Mid/Long threshold')
ax2.set_xticks(range(len(style_history)))
ax2.set_xticklabels(style_history['period'])
ax2.set_ylabel('Duration (Years)')
ax2.set_title('Duration Style by Period')
ax2.legend(fontsize=8)
ax2.grid(True, axis='y', alpha=0.3)

plt.suptitle(f'{FUND_NAME} ({FUND_CODE}) Style Evolution', fontsize=13, fontweight='bold')
plt.tight_layout()

hist_path = os.path.join(output_dir, f'{FUND_CODE}_style_evolution.png')
plt.savefig(hist_path, dpi=150, bbox_inches='tight')
plt.show()
print(f'\u98ce格测试图已保存: {hist_path}')
```
