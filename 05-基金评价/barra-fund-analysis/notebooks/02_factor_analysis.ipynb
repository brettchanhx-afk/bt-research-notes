# 02 - 因子构建与分析

本 Notebook 用于：
1. 构建 Barra 风格因子
2. 因子相关性分析
3. 因子描述性统计

```python
import sys
sys.path.insert(0, '..')

import config
from barra import DataLoader, FactorBuilder

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False
```

## 1. 加载数据

```python
loader = DataLoader(config.DATA_DIR, config.START_DATE, config.END_DATE)
index_df = loader.load_all_indexes(config.FACTOR_INDEXES)

print("原始指数收益率:")
index_df.head()
```

## 2. 构建 Barra 因子

```python
builder = FactorBuilder(index_df)
factors = builder.build()

print("Barra 风格因子:")
print(factors.head(10))
```

## 3. 因子描述性统计

```python
stats = FactorBuilder.describe(factors)
print(stats)
```

```python
# 因子分布直方图
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for idx, col in enumerate(factors.columns):
    ax = axes[idx]
    factors[col].hist(bins=50, ax=ax, alpha=0.7)
    ax.set_title(f'{col} 分布')
    ax.axvline(factors[col].mean(), color='red', linestyle='--', label='均值')
    ax.legend()

plt.tight_layout()
plt.show()
```

## 4. 因子相关性

```python
corr = factors.corr()

fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdBu_r', center=0, ax=ax,
            square=True, cbar_kws={'label': '相关系数'})
ax.set_title('Barra 因子相关性矩阵')
plt.show()
```

## 5. 因子累计收益

```python
cum_factors = (1 + factors).cumprod()

fig, ax = plt.subplots(figsize=(12, 5))
for col in cum_factors.columns:
    ax.plot(cum_factors.index, cum_factors[col], label=col)
ax.set_title('风格因子累计收益')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```
